Create an experiment

```python
from pathlib import Path

import IPython.core.display as core_display
from IPython.display import HTML, display
from labml import lab

# labml still imports display/HTML from IPython.core.display in notebook mode.
# Newer IPython versions expose them from IPython.display instead.
core_display.display = display
core_display.HTML = HTML

# Avoid relying on .labml.yaml when running this notebook directly.
lab.configure({"path": str(Path.cwd()), "app_url": None})

experiment.create(name="diffuse", writers={"screen"})
```

Configurations

```python
configs = Configs()
```

```python
experiment.configs(configs, {
    "dataset": "MNIST",
    "image_channels": 1,
    "epochs": 5,
})
```

Initialize

```python
configs.init()
```

```python
with experiment.start():
    configs.run()
```
